## Exercise 1: Create a Quantum State as a Column Vector

### Background

In quantum mechanics, a quantum state is represented as a column vector in a complex vector space. For a single qubit, we use a 2D column vector:

$$|\psi\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$$

where $|\alpha|^2 + |\beta|^2 = 1$ (normalisation condition).

The computational basis states are:
- $|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$
- $|1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$

### Task

Create the quantum state $|0\rangle$ as a NumPy column vector.

```python
# SOLUTION: Create quantum state |0⟩ as a column vector
quantum_state = np.array([[1], [0]])

print("Quantum state |0⟩:")
print(quantum_state)
print(f"\nShape: {quantum_state.shape}")
print(f"Type: Column vector with shape (2, 1)")

# Verify normalisation
norm = np.linalg.norm(quantum_state)
print(f"\nNorm: {norm:.4f}")
assert np.isclose(norm, 1.0), "State must be normalised!"
print("✓ State is properly normalised")
```

### Explanation

We use a **column vector** `[[1], [0]]` rather than a row vector `[1, 0]` because:
1. Matrix multiplication requires column vectors for quantum states
2. Shape `(2, 1)` enables proper broadcasting with gate matrices
3. Matches standard quantum mechanics notation

The normalisation condition $\langle\psi|\psi\rangle = 1$ ensures the probabilities sum to 1.

## Exercise 2: Create the Pauli-X Gate Matrix

### Background

The Pauli-X gate is the quantum equivalent of a classical NOT gate. It flips the computational basis states:

$$X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$$

Properties:
- $X|0\rangle = |1\rangle$
- $X|1\rangle = |0\rangle$
- $X^2 = I$ (applying X twice returns to original state)

### Task

Create the Pauli-X gate as a 2×2 NumPy matrix.

```python
# SOLUTION: Create Pauli-X gate matrix
pauli_x_gate = np.array([[0, 1], 
                         [1, 0]])

print("Pauli-X (NOT) Gate:")
print(pauli_x_gate)
print(f"\nShape: {pauli_x_gate.shape}")

# Verify it's unitary: X† X = I
x_dagger = pauli_x_gate.conj().T
product = np.dot(x_dagger, pauli_x_gate)
print(f"\nX† X (should be identity):")
print(product)
assert np.allclose(product, np.eye(2)), "X gate must be unitary!"
print("✓ Gate is unitary")
```

### Explanation

The Pauli-X gate is **unitary**, meaning $X^\dagger X = I$ where $X^\dagger$ is the conjugate transpose.

Unitary gates preserve the normalisation of quantum states and are reversible - key requirements for quantum computation.

## Exercise 3: Apply the Pauli-X Gate

### Background

To apply a quantum gate to a state, we perform matrix-vector multiplication:

$$|\psi'\rangle = U|\psi\rangle$$

For the Pauli-X gate:

$$X|0\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \end{pmatrix} = |1\rangle$$

### Task

Apply the Pauli-X gate to the quantum state $|0\rangle$.

```python
# SOLUTION: Apply Pauli-X gate to quantum state
transformed_state = np.dot(pauli_x_gate, quantum_state)

print("Original state |0⟩:")
print(quantum_state.flatten())

print("\nAfter applying X gate:")
print(transformed_state.flatten())

print("\nResult: |0⟩ → |1⟩")

# Verify the result is |1⟩
expected_state_1 = np.array([[0], [1]])
assert np.allclose(transformed_state, expected_state_1), "Result should be |1⟩"
print("✓ Correctly flipped |0⟩ to |1⟩")
```

### Explanation

The matrix multiplication `np.dot(pauli_x_gate, quantum_state)` computes:

$$\begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \cdot 1 + 1 \cdot 0 \\ 1 \cdot 1 + 0 \cdot 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

This demonstrates the bit-flip operation of the X gate.

## Exercise 4: Create the Hadamard Gate

### Background

The Hadamard gate creates quantum superposition:

$$H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$

Properties:
- $H|0\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle) = |+\rangle$
- $H|1\rangle = \frac{1}{\sqrt{2}}(|0\rangle - |1\rangle) = |-\rangle$
- $H^2 = I$ (Hadamard is its own inverse)

### Task

Create the Hadamard gate matrix.

```python
# SOLUTION: Create Hadamard gate matrix
hadamard_gate = np.array([[1, 1], 
                          [1, -1]]) / np.sqrt(2)

print("Hadamard Gate:")
print(hadamard_gate)
print(f"\nNumerical values:")
print(f"1/√2 ≈ {1/np.sqrt(2):.4f}")

# Verify it's unitary
h_dagger = hadamard_gate.conj().T
product = np.dot(h_dagger, hadamard_gate)
print(f"\nH† H (should be identity):")
print(product)
assert np.allclose(product, np.eye(2)), "H gate must be unitary!"
print("✓ Gate is unitary")

# Verify H² = I
h_squared = np.dot(hadamard_gate, hadamard_gate)
print(f"\nH² (should be identity):")
print(h_squared)
assert np.allclose(h_squared, np.eye(2)), "H² should equal I"
print("✓ H² = I (self-inverse property)")
```

### Explanation

The Hadamard gate is fundamental in quantum computing for creating **superposition**. The factor $\frac{1}{\sqrt{2}}$ ensures normalisation.

The self-inverse property ($H^2 = I$) means applying Hadamard twice returns the original state.

## Exercise 5: Apply the Hadamard Gate

### Background

Applying the Hadamard gate to $|0\rangle$ creates an equal superposition:

$$H|0\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix} = |+\rangle$$

This state has equal probability (50%) of measuring 0 or 1.

### Task

Apply the Hadamard gate to $|0\rangle$.

```python
# SOLUTION: Apply Hadamard gate to quantum state
transformed_state_hadamard = np.dot(hadamard_gate, quantum_state)

print("Original state |0⟩:")
print(quantum_state.flatten())

print("\nAfter applying Hadamard gate:")
print(transformed_state_hadamard.flatten())

print("\nThis is the |+⟩ state: (|0⟩ + |1⟩)/√2")

# Calculate measurement probabilities
prob_0 = np.abs(transformed_state_hadamard[0, 0])**2
prob_1 = np.abs(transformed_state_hadamard[1, 0])**2

print(f"\nMeasurement probabilities:")
print(f"P(0) = |α|² = {prob_0:.4f} (50%)")
print(f"P(1) = |β|² = {prob_1:.4f} (50%)")

# Verify equal superposition
assert np.isclose(prob_0, 0.5), "Should have 50% probability for |0⟩"
assert np.isclose(prob_1, 0.5), "Should have 50% probability for |1⟩"
print("\n✓ Equal superposition created")
```

### Explanation

The Hadamard gate transforms the definite state $|0\rangle$ into a **superposition state** $|+\rangle$.

In superposition:
- Before measurement: The qubit is in both states simultaneously
- Upon measurement: Collapses to $|0\rangle$ or $|1\rangle$ with equal probability (50%)

This is fundamentally different from classical bits, which are always in a definite state.

## Exercise 6: Calculate Inner Product

### Background

The inner product (also called dot product or scalar product) between two quantum states $|\psi\rangle$ and $|\phi\rangle$ is:

$$\langle\psi|\phi\rangle = \sum_i \psi_i^* \phi_i$$

where $\psi_i^*$ denotes the complex conjugate.

**Physical meaning:**
- $|\langle\psi|\phi\rangle|^2$ gives the probability of measuring $|\psi\rangle$ when the system is in $|\phi\rangle$
- For orthogonal states (like $|0\rangle$ and $|1\rangle$): $\langle 0|1\rangle = 0$

### Task

Calculate the inner product between $|0\rangle$ and $|1\rangle$.

```python
# SOLUTION: Calculate inner product of two quantum states
quantum_state_2 = np.array([[0], [1]])  # |1⟩

# Use np.vdot for proper complex conjugation
inner_product = np.vdot(quantum_state.flatten(), quantum_state_2.flatten())

print("State 1 (|0⟩):", quantum_state.flatten())
print("State 2 (|1⟩):", quantum_state_2.flatten())
print(f"\nInner product ⟨0|1⟩ = {inner_product}")

# Verify orthogonality
assert np.isclose(inner_product, 0), "States |0⟩ and |1⟩ should be orthogonal"
print("\n✓ States are orthogonal (inner product = 0)")

# Demonstrate with non-orthogonal states
plus_state = hadamard_gate @ quantum_state  # |+⟩
inner_product_plus_0 = np.vdot(plus_state.flatten(), quantum_state.flatten())
print(f"\nFor comparison, ⟨+|0⟩ = {inner_product_plus_0:.4f}")
print(f"Probability of measuring |+⟩ as |0⟩: {np.abs(inner_product_plus_0)**2:.4f} (50%)")
```

### Explanation

The computational basis states $|0\rangle$ and $|1\rangle$ are **orthogonal**: $\langle 0|1\rangle = 0$.

This means:
- They represent completely distinguishable states
- Measuring $|0\rangle$ can never yield outcome "1"
- They form an orthonormal basis for the 2D Hilbert space

**Important:** Use `np.vdot()` rather than `np.dot()` for inner products, as it properly handles complex conjugation.

## Exercise 7: Create Gate Sequence (Quantum Circuit)

### Background

Multiple gates can be applied sequentially to create quantum circuits. The order matters!

For gates $U_1, U_2, U_3$ applied in sequence:

$$|\psi'\rangle = U_3(U_2(U_1|\psi\rangle)) = (U_3 U_2 U_1)|\psi\rangle$$

Note the reverse order in the matrix product.

### Task

Apply the gate sequence X → H to $|0\rangle$ (first X, then H).

```python
# SOLUTION: Create quantum circuit with gate sequence X-H
# Apply X first, then H: H(X|0⟩)
circuit_result = np.dot(hadamard_gate, np.dot(pauli_x_gate, quantum_state))

print("Gate sequence: X → H")
print("\nStep by step:")

# Step 1: Apply X
after_x = np.dot(pauli_x_gate, quantum_state)
print(f"1. X|0⟩ = |1⟩:")
print(f"   {after_x.flatten()}")

# Step 2: Apply H
after_h = np.dot(hadamard_gate, after_x)
print(f"\n2. H|1⟩ = (|0⟩ - |1⟩)/√2 = |-⟩:")
print(f"   {after_h.flatten()}")

print("\nFinal result:")
print(circuit_result.flatten())

# This is the |-⟩ state
minus_state = np.array([[1], [-1]]) / np.sqrt(2)
assert np.allclose(circuit_result, minus_state), "Should produce |-⟩ state"
print("\n✓ Successfully created |-⟩ state via X→H sequence")

# Alternative: Pre-compute combined gate
combined_gate = np.dot(hadamard_gate, pauli_x_gate)
circuit_result_alt = np.dot(combined_gate, quantum_state)
assert np.allclose(circuit_result, circuit_result_alt), "Both methods should give same result"
print("✓ Gate composition verified")
```

### Explanation

**Gate order matters!**

- X → H produces $|-\rangle = \frac{1}{\sqrt{2}}(|0\rangle - |1\rangle)$
- H → X would produce a different result

**Two equivalent ways to compute:**
1. Sequential application: `H(X|ψ⟩)`
2. Pre-compose gates: `(HX)|ψ⟩`

Both are mathematically equivalent due to associativity of matrix multiplication.

## Exercise 8: Calculate Tensor Product

### Background

The **tensor product** (Kronecker product) combines quantum states of multiple qubits:

$$|\psi\rangle \otimes |\phi\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix} \otimes \begin{pmatrix} \gamma \\ \delta \end{pmatrix} = \begin{pmatrix} \alpha\gamma \\ \alpha\delta \\ \beta\gamma \\ \beta\delta \end{pmatrix}$$

For example:
$$|0\rangle \otimes |1\rangle = |01\rangle = \begin{pmatrix} 0 \\ 1 \\ 0 \\ 0 \end{pmatrix}$$

### Task

Calculate the tensor product of $|0\rangle$ and $|1\rangle$ to create the two-qubit state $|01\rangle$.

```python
# SOLUTION: Calculate tensor product of two quantum states
tensor_product = np.kron(quantum_state, quantum_state_2)

print("State 1: |0⟩ =", quantum_state.flatten())
print("State 2: |1⟩ =", quantum_state_2.flatten())

print("\nTensor product |0⟩ ⊗ |1⟩ = |01⟩:")
print(tensor_product.flatten())
print(f"\nShape: {tensor_product.shape} (4-dimensional for 2 qubits)")

# Verify the result
expected_01 = np.array([[0], [1], [0], [0]])
assert np.allclose(tensor_product, expected_01), "Should produce |01⟩ state"
print("\n✓ Correctly created two-qubit state |01⟩")

# Demonstrate other two-qubit basis states
print("\n--- All two-qubit computational basis states ---")
state_0 = quantum_state
state_1 = quantum_state_2

basis_00 = np.kron(state_0, state_0)
basis_01 = np.kron(state_0, state_1)
basis_10 = np.kron(state_1, state_0)
basis_11 = np.kron(state_1, state_1)

print(f"|00⟩ = {basis_00.flatten()}")
print(f"|01⟩ = {basis_01.flatten()}")
print(f"|10⟩ = {basis_10.flatten()}")
print(f"|11⟩ = {basis_11.flatten()}")
```

### Explanation

The tensor product expands the state space:
- 1 qubit: 2-dimensional space (2¹ states)
- 2 qubits: 4-dimensional space (2² states)
- n qubits: 2ⁿ-dimensional space

**Computational basis for 2 qubits:**
- $|00\rangle = (1, 0, 0, 0)^T$
- $|01\rangle = (0, 1, 0, 0)^T$
- $|10\rangle = (0, 0, 1, 0)^T$
- $|11\rangle = (0, 0, 0, 1)^T$

These four vectors form an orthonormal basis for the 4D Hilbert space of two qubits.

## Exercise 9: Create Identity Matrix

### Background

The **identity gate** is the "do nothing" operation:

$$I = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix}$$

Properties:
- $I|\psi\rangle = |\psi\rangle$ for any state $|\psi\rangle$
- $IU = UI = U$ for any gate $U$
- Useful in multi-qubit gates (e.g., $I \otimes X$ applies X to second qubit only)

### Task

Create the 2×2 identity matrix.

```python
# SOLUTION: Create identity matrix
identity_matrix = np.eye(2)

print("Identity matrix (2×2):")
print(identity_matrix)

# Verify it leaves states unchanged
result_0 = np.dot(identity_matrix, quantum_state)
result_1 = np.dot(identity_matrix, quantum_state_2)

print("\nI|0⟩ = |0⟩:", result_0.flatten())
print("I|1⟩ = |1⟩:", result_1.flatten())

assert np.allclose(result_0, quantum_state), "Should leave |0⟩ unchanged"
assert np.allclose(result_1, quantum_state_2), "Should leave |1⟩ unchanged"
print("\n✓ Identity gate correctly preserves states")

# Demonstrate use in multi-qubit gates
# I ⊗ X applies X to second qubit only
i_tensor_x = np.kron(identity_matrix, pauli_x_gate)
print("\n--- Multi-qubit application: I ⊗ X ---")
print("This gate applies X to the second qubit only")

test_state_00 = np.kron(quantum_state, quantum_state)
result = i_tensor_x @ test_state_00
expected_01 = np.kron(quantum_state, quantum_state_2)

print(f"(I ⊗ X)|00⟩ = |01⟩: {result.flatten()}")
assert np.allclose(result, expected_01), "Should flip only second qubit"
print("✓ Multi-qubit gate works correctly")
```

### Explanation

The identity matrix is fundamental:

1. **As a gate:** Does nothing to quantum states
2. **In linear algebra:** Multiplicative identity ($AI = IA = A$)
3. **In multi-qubit systems:** Allows selective gate application

Example: $I \otimes X$ creates a 4×4 matrix that:
- Leaves the first qubit alone (identity)
- Flips the second qubit (X gate)

## Exercise 10: Calculate Eigenvalues and Eigenvectors

### Background

Eigenvalue decomposition is central to quantum mechanics. For a matrix $A$:

$$A|\lambda\rangle = \lambda|\lambda\rangle$$

where:
- $\lambda$ is an eigenvalue (scalar)
- $|\lambda\rangle$ is the corresponding eigenvector

**Physical interpretation:**
- Eigenvalues = possible measurement outcomes
- Eigenvectors = states that yield definite measurement results
- Observable operators are Hermitian matrices with real eigenvalues

### Task

Find eigenvalues and eigenvectors of the matrix $A = \begin{pmatrix} 3 & 1 \\ 1 & 2 \end{pmatrix}$.

```python
# SOLUTION: Calculate eigenvalues and eigenvectors
matrix = np.array([[3, 1], 
                   [1, 2]])

print("Matrix A:")
print(matrix)

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(matrix)

print("\nEigenvalues:")
for i, eigenvalue in enumerate(eigenvalues):
    print(f"  λ_{i+1} = {eigenvalue:.4f}")

print("\nEigenvectors (as columns):")
print(eigenvectors)

# Verify the eigenvalue equation for each eigenpair
print("\n--- Verification: A|λ⟩ = λ|λ⟩ ---")
for i in range(len(eigenvalues)):
    eigenvector = eigenvectors[:, i]
    eigenvalue = eigenvalues[i]
    
    # Left side: A|λ⟩
    left_side = matrix @ eigenvector
    
    # Right side: λ|λ⟩
    right_side = eigenvalue * eigenvector
    
    print(f"\nEigenpair {i+1}:")
    print(f"  A|λ_{i+1}⟩ = {left_side}")
    print(f"  λ_{i+1}|λ_{i+1}⟩ = {right_side}")
    
    assert np.allclose(left_side, right_side), f"Eigenvalue equation failed for pair {i+1}"
    print(f"  ✓ Verified: A|λ_{i+1}⟩ = λ_{i+1}|λ_{i+1}⟩")

# Check orthogonality of eigenvectors
inner_prod = np.dot(eigenvectors[:, 0], eigenvectors[:, 1])
print(f"\nInner product of eigenvectors: {inner_prod:.10f}")
if np.isclose(inner_prod, 0):
    print("✓ Eigenvectors are orthogonal")
else:
    print("Note: Eigenvectors are not orthogonal (matrix is not Hermitian)")

print("\n✓ Eigenvalue decomposition complete")
```

### Explanation

**Eigenvalue decomposition in quantum mechanics:**

1. **Measurement operators (observables)** are represented by Hermitian matrices
2. **Eigenvalues** are the possible measurement outcomes (always real for Hermitian operators)
3. **Eigenvectors** are the quantum states that give definite measurements

**Example:** The Pauli-Z operator:
$$Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

has eigenvalues $+1$ and $-1$ with eigenvectors $|0\rangle$ and $|1\rangle$. Measuring a qubit in state $|0\rangle$ along the Z-axis always yields $+1$.

**Spectral decomposition:** Any Hermitian operator can be written as:
$$A = \sum_i \lambda_i |\lambda_i\rangle\langle\lambda_i|$$

This is the foundation of quantum measurement theory!
