## Exercise 1: Define Rank Basis States

### Background

We'll represent 8 card ranks (7, 8, 9, 10, Jack, Queen, King, Ace) as basis states in an 8-dimensional Hilbert space.

### Implementation

Each rank is a column vector with a 1 in one position and 0s elsewhere.

```python
# SOLUTION: Define 8 rank basis states

# Create rank basis states (8D column vectors)
rank_7 = np.array([[1], [0], [0], [0], [0], [0], [0], [0]])
rank_8 = np.array([[0], [1], [0], [0], [0], [0], [0], [0]])
rank_9 = np.array([[0], [0], [1], [0], [0], [0], [0], [0]])
rank_10 = np.array([[0], [0], [0], [1], [0], [0], [0], [0]])
rank_J = np.array([[0], [0], [0], [0], [1], [0], [0], [0]])
rank_Q = np.array([[0], [0], [0], [0], [0], [1], [0], [0]])
rank_K = np.array([[0], [0], [0], [0], [0], [0], [1], [0]])
rank_A = np.array([[0], [0], [0], [0], [0], [0], [0], [1]])

# Store in a list for easy access
ranks = [rank_7, rank_8, rank_9, rank_10, rank_J, rank_Q, rank_K, rank_A]
rank_names = ['7', '8', '9', '10', 'J', 'Q', 'K', 'A']

print("Rank basis states created:")
for name, rank in zip(rank_names, ranks):
    print(f"  |{name}⟩: shape {rank.shape}, norm {np.linalg.norm(rank):.1f}")

# Verify orthonormality
print("\nOrthonormality check:")
print("  ⟨7|7⟩ =", np.dot(rank_7.T, rank_7)[0, 0])
print("  ⟨7|8⟩ =", np.dot(rank_7.T, rank_8)[0, 0])
print("  ⟨Q|K⟩ =", np.dot(rank_Q.T, rank_K)[0, 0])
```

**Expected:** All ranks have norm 1, orthogonal ranks have inner product 0

## Exercise 2: Define Suit Basis States

### Background

We'll represent 4 suits (♠, ♥, ♦, ♣) as basis states in a 4-dimensional Hilbert space.

### Implementation

```python
# SOLUTION: Define 4 suit basis states

# Create suit basis states (4D column vectors)
spades = np.array([[1], [0], [0], [0]])
hearts = np.array([[0], [1], [0], [0]])
diamonds = np.array([[0], [0], [1], [0]])
clubs = np.array([[0], [0], [0], [1]])

# Store in a list for easy access
suits = [spades, hearts, diamonds, clubs]
suit_names = ['♠', '♥', '♦', '♣']

print("Suit basis states created:")
for name, suit in zip(suit_names, suits):
    print(f"  |{name}⟩: shape {suit.shape}, norm {np.linalg.norm(suit):.1f}")

# Verify orthonormality
print("\nOrthonormality check:")
print("  ⟨♠|♠⟩ =", np.dot(spades.T, spades)[0, 0])
print("  ⟨♠|♥⟩ =", np.dot(spades.T, hearts)[0, 0])
print("  ⟨♦|♣⟩ =", np.dot(diamonds.T, clubs)[0, 0])
```

**Expected:** All suits have norm 1, orthogonal suits have inner product 0

## Exercise 3: Construct Card States Using Kronecker Product

### Background

A complete card state is the **tensor product** of a rank state and a suit state:

$$|\text{card}\rangle = |\text{rank}\rangle \otimes |\text{suit}\rangle$$

In NumPy: `np.kron(rank, suit)`

### Implementation

```python
# SOLUTION: Construct card states

# Example: 7 of Spades
card_7_spades = np.kron(rank_7, spades)

print("7 of Spades:")
print(f"  Shape: {card_7_spades.shape}")
print(f"  Norm: {np.linalg.norm(card_7_spades):.1f}")
print(f"  Non-zero element at index: {np.argmax(np.abs(card_7_spades))}")
print()

# Create a few more example cards
card_A_hearts = np.kron(rank_A, hearts)
card_K_diamonds = np.kron(rank_K, diamonds)
card_Q_clubs = np.kron(rank_Q, clubs)

print("Additional cards created:")
print(f"  Ace of Hearts: shape {card_A_hearts.shape}")
print(f"  King of Diamonds: shape {card_K_diamonds.shape}")
print(f"  Queen of Clubs: shape {card_Q_clubs.shape}")
print()

# Verify cards are orthonormal
print("Orthogonality check:")
inner_same = np.dot(card_7_spades.T, card_7_spades)[0, 0]
inner_diff = np.dot(card_7_spades.T, card_A_hearts)[0, 0]
print(f"  ⟨7♠|7♠⟩ = {inner_same:.10f} (should be 1)")
print(f"  ⟨7♠|A♥⟩ = {inner_diff:.10f} (should be 0)")
```

**Expected:** All cards are 32D vectors with norm 1, different cards are orthogonal

### Visualise Card State

```python
# SOLUTION: Visualise a card state

fig, ax = plt.subplots(figsize=(14, 4))

indices = np.arange(32)
amplitudes = card_7_spades.flatten()

ax.bar(indices, amplitudes, color='steelblue', edgecolor='black', linewidth=1.5)
ax.set_xlabel('Basis State Index', fontsize=12, fontweight='bold')
ax.set_ylabel('Amplitude', fontsize=12, fontweight='bold')
ax.set_title('7 of Spades - Quantum State Representation', fontsize=14, fontweight='bold')
ax.set_xlim([-1, 32])
ax.grid(axis='y', alpha=0.3)

# Mark the non-zero element
non_zero_idx = np.argmax(np.abs(amplitudes))
ax.axvline(x=non_zero_idx, color='red', linestyle='--', linewidth=2, 
           label=f'Non-zero at index {non_zero_idx}')
ax.legend(fontsize=12)

plt.tight_layout()
plt.show()

print(f"The 7 of Spades has amplitude 1 at index {non_zero_idx} and 0 everywhere else.")
```

## Exercise 4: Create All 32 Card States

### Your Task

Generate all 32 possible card states systematically.

### Implementation

```python
# SOLUTION: Create all 32 card states

# Generate all card states
all_cards = {}
all_card_names = []

for rank_name, rank_state in zip(rank_names, ranks):
    for suit_name, suit_state in zip(suit_names, suits):
        card_name = f"{rank_name}{suit_name}"
        card_state = np.kron(rank_state, suit_state)
        all_cards[card_name] = card_state
        all_card_names.append(card_name)

print(f"Created {len(all_cards)} card states:")
print()

# Display cards in a nice format
print("All cards (8 ranks × 4 suits = 32 cards):")
for i, card_name in enumerate(all_card_names):
    print(f"  {card_name:4s}", end='')
    if (i + 1) % 8 == 0:
        print()  # New line after every 8 cards

# Verify all are orthonormal
print("\nVerifying orthonormality...")
errors = []
for i, card1_name in enumerate(all_card_names):
    for j, card2_name in enumerate(all_card_names):
        inner = np.dot(all_cards[card1_name].T, all_cards[card2_name])[0, 0]
        expected = 1.0 if i == j else 0.0
        error = abs(inner - expected)
        if error > 1e-10:
            errors.append((card1_name, card2_name, error))

if len(errors) == 0:
    print("✓ All cards are orthonormal!")
else:
    print(f"⚠ Found {len(errors)} orthonormality violations")
    for c1, c2, err in errors[:5]:
        print(f"    ⟨{c1}|{c2}⟩ error: {err}")
```

**Expected:** 32 orthonormal card states

## Exercise 5: Calculate Inner Products Between Cards

### Background

The inner product $\langle \psi | \phi \rangle$ measures the "overlap" between two quantum states.

### Implementation

```python
# SOLUTION: Calculate inner products

# Select some example cards
card1 = all_cards['7♠']
card2 = all_cards['7♠']  # Same card
card3 = all_cards['A♥']  # Different card
card4 = all_cards['7♥']  # Same rank, different suit

# Calculate inner products
inner_same = np.dot(card1.T.conj(), card2)[0, 0]
inner_diff = np.dot(card1.T.conj(), card3)[0, 0]
inner_rank = np.dot(card1.T.conj(), card4)[0, 0]

print("Inner Product Calculations:")
print(f"  ⟨7♠|7♠⟩ = {inner_same:.10f} (same card → 1)")
print(f"  ⟨7♠|A♥⟩ = {inner_diff:.10f} (different card → 0)")
print(f"  ⟨7♠|7♥⟩ = {inner_rank:.10f} (same rank, different suit → 0)")
print()

# Create inner product matrix for all cards
print("Computing full inner product matrix...")
n_cards = len(all_card_names)
inner_product_matrix = np.zeros((n_cards, n_cards))

for i, card_i in enumerate(all_card_names):
    for j, card_j in enumerate(all_card_names):
        inner = np.dot(all_cards[card_i].T.conj(), all_cards[card_j])[0, 0]
        inner_product_matrix[i, j] = np.abs(inner)

print("✓ Inner product matrix computed (32×32)")
```

### Visualise Inner Product Matrix

```python
# Visualise inner product matrix as heatmap
fig, ax = plt.subplots(figsize=(12, 10))

im = ax.imshow(inner_product_matrix, cmap='Blues', interpolation='nearest')
ax.set_title('Inner Product Matrix for All Cards', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Card Index', fontsize=12, fontweight='bold')
ax.set_ylabel('Card Index', fontsize=12, fontweight='bold')

# Add colorbar
cbar = plt.colorbar(im, ax=ax)
cbar.set_label('|⟨card_i|card_j⟩|', fontsize=12, fontweight='bold')

# Add grid
ax.set_xticks(np.arange(0, 32, 4))
ax.set_yticks(np.arange(0, 32, 4))
ax.grid(which='major', color='gray', linestyle='-', linewidth=0.5, alpha=0.3)

plt.tight_layout()
plt.show()

print("The diagonal shows 1 (perfect overlap with self).")
print("Off-diagonal elements are 0 (cards are orthogonal).")
```

**Expected:** Identity matrix pattern (1 on diagonal, 0 elsewhere)

## Exercise 6: Calculate Measurement Probabilities

### Background

The probability of measuring card $|\psi\rangle$ when the system is in state $|\phi\rangle$ is:

$$P(\psi|\phi) = |\langle \psi | \phi \rangle|^2$$

### Implementation

```python
# SOLUTION: Calculate measurement probabilities

# Suppose the system is in state |7♠⟩
system_state = all_cards['7♠']

# Calculate probability of measuring each card
probabilities = {}
for card_name in all_card_names:
    card_state = all_cards[card_name]
    inner = np.dot(card_state.T.conj(), system_state)[0, 0]
    prob = np.abs(inner)**2
    probabilities[card_name] = prob

# Display top probabilities
sorted_probs = sorted(probabilities.items(), key=lambda x: x[1], reverse=True)

print("Measurement probabilities when system is in state |7♠⟩:")
print()
for card_name, prob in sorted_probs[:10]:
    print(f"  P({card_name}) = {prob:.10f}")
print("  ...")
print()

# Verify probabilities sum to 1
total_prob = sum(probabilities.values())
print(f"Total probability: {total_prob:.10f} (should be 1.0)")

# As expected, only |7♠⟩ has probability 1
```

**Expected:** P(7♠) = 1.0, all others = 0.0

## Exercise 7: Create Superposition States

### Background

A **superposition** is a linear combination of basis states:

$$|\psi\rangle = \alpha|\text{card}_1\rangle + \beta|\text{card}_2\rangle$$

where $|\alpha|^2 + |\beta|^2 = 1$ (normalization).

### Implementation

```python
# SOLUTION: Create superposition states

# Example 1: Equal superposition of two cards
card_A = all_cards['7♠']
card_B = all_cards['A♥']

# Create superposition: (|7♠⟩ + |A♥⟩)/√2
superposition_AB = (card_A + card_B) / np.sqrt(2)

print("Superposition 1: (|7♠⟩ + |A♥⟩)/√2")
print(f"  Shape: {superposition_AB.shape}")
print(f"  Norm: {np.linalg.norm(superposition_AB):.10f} (should be 1)")
print()

# Calculate measurement probabilities
prob_7S = np.abs(np.dot(card_A.T.conj(), superposition_AB)[0, 0])**2
prob_AH = np.abs(np.dot(card_B.T.conj(), superposition_AB)[0, 0])**2
prob_other = np.abs(np.dot(all_cards['K♦'].T.conj(), superposition_AB)[0, 0])**2

print("Measurement probabilities:")
print(f"  P(7♠) = {prob_7S:.4f} (should be 0.5)")
print(f"  P(A♥) = {prob_AH:.4f} (should be 0.5)")
print(f"  P(K♦) = {prob_other:.10f} (should be 0)")
print()

# Example 2: Unequal superposition
alpha = np.sqrt(0.7)
beta = np.sqrt(0.3)

superposition_unequal = alpha * card_A + beta * card_B

print("Superposition 2: 0.7|7♠⟩ + 0.3|A♥⟩ (unnormalized coefficients)")
print(f"  Norm: {np.linalg.norm(superposition_unequal):.10f}")

prob_7S_2 = np.abs(np.dot(card_A.T.conj(), superposition_unequal)[0, 0])**2
prob_AH_2 = np.abs(np.dot(card_B.T.conj(), superposition_unequal)[0, 0])**2

print(f"  P(7♠) = {prob_7S_2:.4f} (should be 0.7)")
print(f"  P(A♥) = {prob_AH_2:.4f} (should be 0.3)")
```

### Visualise Superposition State

```python
# Visualise superposition amplitudes
fig, ax = plt.subplots(figsize=(14, 4))

indices = np.arange(32)
amplitudes = superposition_AB.flatten()

colors = ['red' if amp != 0 else 'lightgray' for amp in amplitudes]
ax.bar(indices, amplitudes, color=colors, edgecolor='black', linewidth=1.5)
ax.set_xlabel('Basis State Index', fontsize=12, fontweight='bold')
ax.set_ylabel('Amplitude', fontsize=12, fontweight='bold')
ax.set_title('Superposition State: (|7♠⟩ + |A♥⟩)/√2', fontsize=14, fontweight='bold')
ax.set_xlim([-1, 32])
ax.grid(axis='y', alpha=0.3)
ax.axhline(y=0, color='black', linewidth=0.8)

plt.tight_layout()
plt.show()

print("Two non-zero amplitudes correspond to |7♠⟩ and |A♥⟩.")
```

## Exercise 8: Superposition of Multiple Cards

### Your Task

Create a superposition of multiple cards and calculate measurement probabilities.

### Implementation

```python
# SOLUTION: Superposition of 4 cards (all Aces)

# Create equal superposition of all 4 Aces
superposition_aces = (
    all_cards['A♠'] + 
    all_cards['A♥'] + 
    all_cards['A♦'] + 
    all_cards['A♣']
) / 2.0  # Normalization: 1/√4 = 1/2

print("Superposition of all Aces:")
print(f"  Norm: {np.linalg.norm(superposition_aces):.10f} (should be 1)")
print()

# Calculate probabilities
print("Measurement probabilities:")
for suit_name in suit_names:
    card_name = f"A{suit_name}"
    card_state = all_cards[card_name]
    prob = np.abs(np.dot(card_state.T.conj(), superposition_aces)[0, 0])**2
    print(f"  P({card_name}) = {prob:.4f} (should be 0.25)")

# Check probability of non-Ace card
prob_7S = np.abs(np.dot(all_cards['7♠'].T.conj(), superposition_aces)[0, 0])**2
print(f"  P(7♠) = {prob_7S:.10f} (should be 0)")
print()

# Verify total probability
total = sum(np.abs(np.dot(all_cards[name].T.conj(), superposition_aces)[0, 0])**2 
           for name in all_card_names)
print(f"Total probability: {total:.10f}")
```

**Expected:** Each Ace has P = 0.25, all other cards P = 0

## Exercise 9: The Magic Trick

### Background

Now we'll perform the "magic trick": given a target card and a prepared superposition state, calculate the probability of "guessing" the correct card.

### Implementation

```python
# SOLUTION: The Magic Trick

def magic_trick(target_card_name, superposition_cards):
    """
    Perform the magic trick.
    
    Args:
        target_card_name: Name of the target card
        superposition_cards: List of card names in superposition
    
    Returns:
        Probability of measuring the target card
    """
    # Create equal superposition
    n = len(superposition_cards)
    superposition = sum(all_cards[name] for name in superposition_cards) / np.sqrt(n)
    
    # Get target card
    target = all_cards[target_card_name]
    
    # Calculate probability
    inner = np.dot(target.T.conj(), superposition)[0, 0]
    prob = np.abs(inner)**2
    
    return prob

# Example 1: Target is in the superposition
print("Magic Trick Example 1:")
print("Superposition: All Aces")
print("Target: Ace of Hearts")
prob1 = magic_trick('A♥', ['A♠', 'A♥', 'A♦', 'A♣'])
print(f"Probability: {prob1:.4f}")
print()

# Example 2: Target is NOT in the superposition
print("Magic Trick Example 2:")
print("Superposition: All Aces")
print("Target: 7 of Spades")
prob2 = magic_trick('7♠', ['A♠', 'A♥', 'A♦', 'A♣'])
print(f"Probability: {prob2:.10f}")
print()

# Example 3: Custom superposition
print("Magic Trick Example 3:")
print("Superposition: All 7s")
print("Target: 7 of Diamonds")
prob3 = magic_trick('7♦', ['7♠', '7♥', '7♦', '7♣'])
print(f"Probability: {prob3:.4f}")
print()

print("The 'magic' is that if the target is in the superposition,")
print("the probability is 1/n (where n is the number of cards in superposition).")
```

## Exercise 10: Probability Distribution Visualization

### Your Task

Create a comprehensive visualization of measurement probabilities for a superposition state.

### Implementation

```python
# SOLUTION: Visualize probability distribution

# Create a more interesting superposition: all face cards
face_cards = ['J♠', 'J♥', 'J♦', 'J♣', 'Q♠', 'Q♥', 'Q♦', 'Q♣', 'K♠', 'K♥', 'K♦', 'K♣']
n_face = len(face_cards)
superposition_faces = sum(all_cards[name] for name in face_cards) / np.sqrt(n_face)

# Calculate probabilities for all cards
all_probs = []
for card_name in all_card_names:
    card_state = all_cards[card_name]
    inner = np.dot(card_state.T.conj(), superposition_faces)[0, 0]
    prob = np.abs(inner)**2
    all_probs.append(prob)

# Create visualization
fig, ax = plt.subplots(figsize=(16, 6))

indices = np.arange(len(all_card_names))
colors = ['red' if prob > 0.01 else 'lightblue' for prob in all_probs]

bars = ax.bar(indices, all_probs, color=colors, edgecolor='black', linewidth=1)
ax.set_xlabel('Card', fontsize=12, fontweight='bold')
ax.set_ylabel('Measurement Probability', fontsize=12, fontweight='bold')
ax.set_title('Probability Distribution for Face Card Superposition', fontsize=14, fontweight='bold')
ax.set_xticks(indices[::4])  # Show every 4th label
ax.set_xticklabels([all_card_names[i] for i in range(0, len(all_card_names), 4)], rotation=45)
ax.grid(axis='y', alpha=0.3)

# Add horizontal line for expected probability
expected_prob = 1 / n_face
ax.axhline(y=expected_prob, color='green', linestyle='--', linewidth=2, 
          label=f'Expected (1/{n_face} = {expected_prob:.4f})')
ax.legend(fontsize=12)

plt.tight_layout()
plt.show()

print(f"Total probability: {sum(all_probs):.10f}")
print(f"Number of face cards: {n_face}")
print(f"Expected probability per face card: {expected_prob:.4f}")
print(f"Actual probability (sample): {all_probs[all_card_names.index('J♠')]:.4f}")
```
